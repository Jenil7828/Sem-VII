```python
# Sample sentence
sentence = "NLTK is a leading platform for building Python programs to work with human language data! Let's explore tokenization, stemming, and lemmatization :) #NLP"
print('Sentence:', sentence)
```

```python
# Imports and NLTK downloads (run once)
import nltk
# The following downloads small corpora/models used by tokenizers/lemmatizers.
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('omw-1.4')

from nltk.tokenize import TweetTokenizer, WordPunctTokenizer, TreebankWordTokenizer, word_tokenize
from nltk.tokenize import MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag
from nltk.corpus import wordnet

# GPU device detection (use CUDA if available)
try:
    import torch
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print('device =', device)
except Exception:
    device = 'cpu'
    print('torch not installed; using CPU')

# Helper: map Treebank POS tags to WordNet POS for lemmatization
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN
```

```python
# 1) Tokenization examples
# Whitespace tokenization
ws_tokens = sentence.split()

# Punctuation-based tokenization (WordPunct)
wp = WordPunctTokenizer()
wp_tokens = wp.tokenize(sentence)

# Treebank (handles contractions)
tb = TreebankWordTokenizer()
tb_tokens = tb.tokenize(sentence)
# Alternatively: word_tokenize uses the Treebank tokenizer under the hood
treebank_tokens = word_tokenize(sentence)

# Tweet tokenizer (handles hashtags, emoticons)
tt = TweetTokenizer()
tweet_tokens = tt.tokenize(sentence)

# MWE tokenization: group ('human','language') into a single token
mwe = MWETokenizer([('human', 'language')])
mwe_input = word_tokenize(sentence)
mwe_tokens = mwe.tokenize(mwe_input)

# Display results
print('Whitespace tokens:', ws_tokens)
print('WordPunct tokens:', wp_tokens)
print('Treebank tokens (TreebankWordTokenizer):', tb_tokens)
print('Treebank tokens (word_tokenize):', treebank_tokens)
print('Tweet tokens:', tweet_tokens)
print('MWE tokens (human_language grouped):', mwe_tokens)
```

```python
# 2) Stemming (Porter vs Snowball)
porter = PorterStemmer()
snowball = SnowballStemmer('english')

# Prepare a list of candidate words (filter out pure punctuation)
candidates = [t for t in treebank_tokens if any(c.isalnum() for c in t)]
print('Candidates for stemming:', candidates)

porter_results = [(w, porter.stem(w)) for w in candidates]
snowball_results = [(w, snowball.stem(w)) for w in candidates]

print('Porter Stemmer results:')
for orig, stem in porter_results:
    print(f'{orig} -> {stem}')

print('Snowball Stemmer (English) results:')
for orig, stem in snowball_results:
    print(f'{orig} -> {stem}')
```

```python
# 3) Lemmatization (WordNet)
lemmatizer = WordNetLemmatizer()

# Lemmatize without POS (default assumes nouns)
lemmatized_default = [(w, lemmatizer.lemmatize(w)) for w in candidates]

# POS tagging + POS-aware lemmatization for better results
pos_tags = pos_tag(candidates)
lemmatized_pos = []
for word, tag in pos_tags:
    wn_tag = get_wordnet_pos(tag)
    lem = lemmatizer.lemmatize(word, pos=wn_tag)
    lemmatized_pos.append((word, tag, wn_tag, lem))

print('Lemmatization (default noun):')
for orig, lem in lemmatized_default:
    print(f'{orig} -> {lem}')

print('POS-tagged lemmatization results:')
for orig, tag, wn_tag, lem in lemmatized_pos:
    print(f'{orig} ({tag}) -> {lem} [WN POS: {wn_tag}]')
```

## Conclusion

- Different tokenizers produce different token sets; Treebank and Tweet tokenizers handle contractions and social-media tokens better than naive whitespace splitting.
- Stemmers (Porter, Snowball) reduce words to roots but sometimes produce non-words; Snowball often gives slightly cleaner stems for English.
- Lemmatization (WordNet) returns dictionary base forms; POS-aware lemmatization improves accuracy for verbs/adjectives/adverbs.
